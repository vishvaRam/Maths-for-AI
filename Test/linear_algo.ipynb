```python
import numpy as np
```

### The Dot Product

```python
u = np.array([0.2,0.6,0.7])
v = np.array([0.9,0.42,0.96])

print(f"Manual: {np.sum(u*v)}")
print(f"numpy: {np.dot(u,v)}")
print(f"Operator: { u @v}")
```

### Scalar projection

```python
norm = np.linalg.norm(v)
print(f"Normalize the length of the vector: {norm}")
print(f"Scalar Projection: {np.dot(u,v)/norm}")
```

### Vector projection

```python
print(f"u : {u}")
print(f"v : {v}")
vector_proj = (np.dot(u,v)/norm**2)*v
print(f"Vector Projection: {vector_proj}")
orthogonal_remainder = u - vector_proj
print(f"Orthogonal Remainder{orthogonal_remainder}")
```
