```python
import numpy as np
```

### The Dot Product

```python
u = np.array([0.2,0.6,0.7])
v = np.array([0.9,0.42,0.96])

print(f"Manual: {np.sum(u*v)}")
print(f"numpy: {np.dot(u,v)}")
print(f"Operator: { u @v}")
```

### Scalar projection

```python
norm = np.linalg.norm(v)
print(f"Normalize the length of the vector: {norm}")
print(f"Scalar Projection: {np.dot(u,v)/norm}")
```

### Vector projection

```python
print(f"u : {u}")
print(f"v : {v}")
vector_proj = (np.dot(u,v)/norm**2)*v
print(f"Vector Projection: {vector_proj}")
orthogonal_remainder = u - vector_proj
print(f"Orthogonal Remainder{orthogonal_remainder}")
```

### 3: Vector Norms 

```python
x = np.array([3.0, -4.0, 1.0, -2.0])

l1_m= np.sum(np.abs(x))
print(f"L1_m : {l1_m}")
l1_np = np.linalg.norm(x,ord=1)
print(f"L1_np : {l1_np}")
```

```python
l2_m = np.sqrt(np.sum(x**2))
print(f"L2_m : {l2_m:.3f}")
l2_np=np.linalg.norm(x, ord=2)
print(f"L2_np : {l2_np:.3f}")
```

```python
linf_m=np.max(np.abs(x))
print(f"L inf : {linf_m}")
linf_np=np.linalg.norm(x,ord=np.inf)
print(f"L_inf : {linf_np}")
```
